```python
debt_df= df[df["INDICATOR_LABEL"].str.contains("debt",case=False)]
```

```python
debt_df
```

```python
debt_df.groupby("REF_AREA_LABEL")["OBS_VALUE"].agg(["mean","max","sum"])
```

```python
import matplotlib.pyplot as plt

```

```python

global_trend= debt_df.groupby("TIME_PERIOD")["OBS_VALUE"].mean()
print(global_trend)
```

```python
#line plot
plt.figure(figsize=(12,6))

plt.plot(global_trend, marker="o", markersize=5,)

plt.title("Global Debt Trend")
plt.xlabel("Year")
plt.ylabel("Average Debt")
plt.xticks(rotation=45)
plt.show()
```

```python
#to retrieve the top 10 countries trend of their obs value . 

top10 =debt_df.groupby("REF_AREA_LABEL")["OBS_VALUE"].mean().sort_values(ascending=True).head(10)
print(top10)
                                                                                          
```

```python
#bar plot
plt.figure(figsize=(12, 6))

plt.barh(top10,width=10)
plt.title("Top 10 Debt Countries")
plt.xlabel("Countries")
plt.ylabel("Average Debt")
plt.xlim(0,25)
plt.show()
```

```python
#histogram 

fig, ax = plt.subplots()

hist1 = ax.hist(debt_df[debt_df["TIME_PERIOD"] == 2020]["OBS_VALUE"],bins=10,color="green",alpha=0.5,label="2020")
hist2 = ax.hist(debt_df[debt_df["TIME_PERIOD"] == 2022]["OBS_VALUE"],bins=10,color="blue",alpha=0.5,label="2022")
ax.set_title("Debt Distribution")
ax.set_xlabel("Debt Value")
ax.set_ylabel("Frequency")
ax.legend()
plt.show()
```
