```python
debt_df= df[df["INDICATOR_LABEL"].str.contains("debt",case=False)]
```

```python
debt_df
```

```python
debt_df.groupby("REF_AREA_LABEL")["OBS_VALUE"].agg(["mean","max","sum"])
```

```python
import matplotlib.pyplot as plt

```

```python

global_trend= debt_df.groupby("TIME_PERIOD")["OBS_VALUE"].mean()
print(global_trend)
```

```python
#line plot
plt.figure(figsize=(12,6))
plt.plot(global_trend)
plt.title("Global Debt Trend")
plt.xlabel("Year")
plt.ylabel("Average Debt")

plt.show()
```

```python
#to retrieve the top 10 countries trend of their obs value . 

top10 =debt_df.groupby("REF_AREA_LABEL")["OBS_VALUE"].mean().sort_values(ascending=True).head(10)
print(top10)
                                                                                          
```

```python
#bar plot
plt.figure(figsize=(12, 6))

top10.plot(kind="bar")
plt.title("Top 10 Debt Countries")
plt.xlabel("Countries")
plt.ylabel("Average Debt")
plt.show()
```

```python
#histogram 
plt.figure(figsize=(10,6))
plt.hist(debt_df["OBS_VALUE"],bins=10)
plt.title("Debt Distribution")
plt.xlabel("Debt Value")
plt.show()
```
